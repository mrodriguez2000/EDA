```python
# ASIGNAR LOS NOMBRES QUE SE ENCUENTRAN EN EL ARCHIVO JSON A LAS COLUMNAS DEL DATASET

# INICIALMENTE SE IMPORTAN LAS LIBRERIAS PARA LEER EL ARCHIVO JSON
import json
import os

# LEER EL JSON
with open("../NOMBRE DE COLUMNAS/Nombres.json", "r", encoding="utf-8") as f:
    column_names = json.load(f)
    column_names = list(column_names.values())  # CONVERTIR LOS VALORES DEL JSON EN UNA LISTA

# ASIGNAR LOS NUEVOS NOMBRES
df_salud.columns = column_names

# MOSTRAR DATAFRAME
df_salud
```

**VISUALIZACIÓN DE LOS DATOS**

```python
# VISUALIZAR LA EDAD PROMEDIO DE LOS PACIENTES CON ENFERMEDADES CARDIACAS AGRUPADO POR GENERO
pacientes_con_enfermedades = df_salud[df_salud["Enfermedad_Cardiaca"] == 1]
edad_promedio_por_genero = pacientes_con_enfermedades.groupby("Genero_Paciente")["Edad_Paciente"].mean().reset_index().replace({
    "Genero_Paciente": {
        0: "Mujeres", 
        1: "Hombres"
        }
    })
edad_promedio_por_genero["Edad_Paciente"] = edad_promedio_por_genero["Edad_Paciente"].astype(int)
edad_promedio_por_genero      
```

```python
# SE GENERA UN GRAFICO DE BARRAS DE LA TABLA ANTERIOR UTILIZANDO PLOTLY
grafico_edad_promedio = px.bar(edad_promedio_por_genero, x="Genero_Paciente", y="Edad_Paciente", color="Genero_Paciente",
                              title="Edad Promedio de Pacientes con Enfermedades Cardiacas por Género",
                              labels={"Genero_Paciente": "Género del Paciente", "Edad_Paciente": "Edad Promedio"})
grafico_edad_promedio.show()
```

**CONCLUSIONES:** Se observa en nuestro dataset que la edad promedio de las personas con enfermedades cardiacas se encuentran en un rango entre los 50 a 60 años de edad.

```python
# HISTOGRAMA DE LA EDAD DE LOS PACIENTES
histograma_edad = px.histogram(df_salud, x="Edad_Paciente", nbins=20, title="Distribución de la Edad de los Pacientes",
                               labels={
                                   "Edad_Paciente": "Edad del Paciente",
                                   "Enfermedad_Cardiaca": "Enfermedad Cardíaca"
                                },
color="Enfermedad_Cardiaca")
histograma_edad.show()
```

**CONCLUSIÓN:** En el histograma se puede observar que gran parte de los pacientes en el dataset se encuentran entre los 54 a 60 años de edad, y esos son los que más sufren de enfermedades cardiacas.

```python
# GENERAR UNA MATRIZ DE CORRELACION PARA VER LAS RELACIONES ENTRE LAS VARIABLES CON PLOTLY
matriz_correlacion = df_salud.corr()
fig = px.imshow(
    matriz_correlacion,
    text_auto=True,
    aspect="auto",
    color_continuous_scale="aggrnyl",
    title="Matriz de Correlación",
)

# AÑADIR DIMENSIONES (ANCHO Y ALTO) AL GRÁFICO
fig.update_layout(
    width=1460,
    height=600,
)

fig.show()
```

**CONCLUSIÓN:** Se pueden observar que las variables con mejor relación a determinar si un paciente tiene enfermedad cardiaca son:

* Tipo_Dolor_Pecho

* Frecuencia_Cardiaca_Maxima

* Angina_Ejercicio

* Depresion_ST_Ejercicio

* Numero_Vasos_Filtrados

```python
# SELECCIONAR LAS COLUMNAS MAS RELEVANTES CON SKLEARN
from sklearn.feature_selection import SelectKBest, f_classif

# CON LA X SE VA A ELIMINAR LA COLUMNA OBJETIVO Y CON LA Y SE VA A SELECCIONAR LA COLUMNA OBJETIVO
X = df_salud.drop("Enfermedad_Cardiaca", axis=1)
y = df_salud["Enfermedad_Cardiaca"]

# SE UTILIZA EL METODO SelectKBest PARA SELECCIONAR LAS 6 CARACTERISTICAS MAS RELEVANTES PARA PREDECIR LA ENFERMEDAD CARDIACA
selector = SelectKBest(score_func=f_classif, k=6)

# SE TRANSFORMA LA MATRIZ DE CARACTERISTICAS X Y SE OBTIENEN LOS NOMBRES DE LAS CARACTERISTICAS SELECCIONADAS
X_new = selector.fit_transform(X, y)

# OBTENER LOS NOMBRES DE LAS CARACTERISTICAS SELECCIONADAS
selected_features = X.columns[selector.get_support()]

# SE IMPRIMEN LAS CARACTERISTICAS SELECCIONADAS
print("LOS CAMPOS SELECCIONADOS SON: {} \n".format(selected_features))

# CREAR UN NUEVO DATAFRAME SOLO CON LAS CARACTERÍSTICAS SELECCIONADAS Y LA COLUMNA OBJETIVO
df_salud_seleccionado = df_salud[selected_features.tolist() + ["Enfermedad_Cardiaca"]]
df_salud_seleccionado
```

**POR ÚLTIMO SE VA A EXPORTAR EL DATASET FINAL CON LAS COLUMNAS RENOMBRADAS**
