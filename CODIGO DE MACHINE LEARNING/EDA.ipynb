**IMPORTAR LIBRERIAS Y DATASETS**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# LIBRERIAS DE VISUALIZACION QUE GENERAN GRAFICOS INTERACTIVOS
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.subplots as sp
```

```python
df_conversions = pd.read_csv("../DATASETS/digital_marketing_campaign_dataset.csv")
df_conversions.head(10)
```

**DICCIONARIO DE DATOS**

**CustomerID**          =       ID del cliente

**Age**                 =       Edad del cliente

**Gender**              =       Genero del cliente

**Income**              =       Ingresos $ del cliente

**CampaignChannel**     =       Canal donde se realizó la campaña digital

**CampaignType**        =       Tipo de campaña digital: Conversión, Retención de clientes, etc

**AdSpend**             =       Monto invertido en la campaña de marketing en USD

**ClickThroughRate**    =       Tasa a la que los clientes hacen clic en el contenido de marketing

**ConversionRate**      =       Tasa de conversión

```python
cantidad_conversiones_medio = df_conversions.groupby("CampaignChannel")["Conversion"].sum().sort_values(ascending=False).reset_index()
cantidad_conversiones_medio.rename(columns={
        "Conversion":"Cantidad de conversiones",
        "CampaignChannel": "Canal"
    }, inplace=True)
cantidad_conversiones_medio
```

```python
# SE GENERA UN GRAFICO EN PLOTLY PARA VISUALIZAR LA CANTIDAD DE CONVERSIONES
fig_conversiones = px.bar(cantidad_conversiones_medio, x="Canal", y="Cantidad de conversiones", title="Cantidad de conversiones por canal de marketing",
color="Cantidad de conversiones", color_continuous_scale="Viridis")

# GENERAR ETIQUETAS EN LAS BARRAS
for i, row in cantidad_conversiones_medio.iterrows():
    fig_conversiones.add_annotation(
        x=row["Canal"],
        y=row["Cantidad de conversiones"],
        text=f"{row['Cantidad de conversiones']}"
    )
fig_conversiones.show()
```

**CONCLUSIONES:** Se observa que los medios Referral y PPC son los que más generan conversiones

```python
# CREAR PORCENTAJE DE CONVERSION POR GENERO
porcentaje_conversion_genero = df_conversions.groupby(["Gender"])["Conversion"].sum().reset_index()

# RENOMBRAR COLUMNAS
porcentaje_conversion_genero.rename(columns={
        "Conversion":"Cantidad de conversiones",
        "Gender": "Genero"
    }, inplace=True)
porcentaje_conversion_genero
```

```python
# GENERAR GRAFICO DE BARRAS CON LA TABLA porcentaje_conversion_genero
fig_conversion_genero = px.bar(porcentaje_conversion_genero, x="Genero", y="Cantidad de conversiones", title="Porcentaje de conversiones por género y edad")

# GENERAR ETIQUETAS EN LAS BARRAS
for i, row in porcentaje_conversion_genero.iterrows():
    fig_conversion_genero.add_annotation(
        x=row["Genero"],
        y=row["Cantidad de conversiones"],
        text=f"{row['Cantidad de conversiones']}"
    )

fig_conversion_genero.show()
```

```python
# AÑADIR CONVERSIONES Y NO CONVERSIONES POR GENERO
conversiones_no_conversiones_por_genero = df_conversions.groupby(["Gender", "Conversion"])["Conversion"].count().reset_index(name="Cantidad").replace({
    0: "No Conversion", 
    1: "Conversion"
    }).rename(columns={
    "Gender": "Genero",
    "Conversion": "Conversion",
    "Cantidad": "Cantidad de conversiones y no conversiones"
})
conversiones_no_conversiones_por_genero
```

```python
# CREAR GRAFICO DE BARRAS APILADAS CON LA TABLA conversiones_no_conversiones_por_genero
fig_conversiones_apiladas = px.bar(conversiones_no_conversiones_por_genero, x="Genero", y="Cantidad de conversiones y no conversiones", color="Conversion", title="Cantidad de conversiones y no conversiones por género")

# AÑADIR ETIQUETAS EN LAS BARRAS SIN FLECHAS

for i, row in conversiones_no_conversiones_por_genero.iterrows():
    fig_conversiones_apiladas.add_annotation(
        x=row["Genero"],
        y=row["Cantidad de conversiones y no conversiones"],
        text=f"{row['Cantidad de conversiones y no conversiones']}"
    )

fig_conversiones_apiladas.show()
```
